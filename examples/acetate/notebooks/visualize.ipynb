# Posterior Visualization

This notebook is intentionally plotting-only. It loads results from
stage `06-learn` and regenerates posterior plots without rerunning
training or MCMC.

```python
from pathlib import Path

from bff import PosteriorResults
from bff.plotting import plot_corner, plot_marginals

ROOT = Path('.').resolve()
LEARN_OUTPUTS = ROOT.parent / '06-learn' / 'outputs'

results = PosteriorResults.load(LEARN_OUTPUTS / 'posterior.pt')
results.prepare_samples()
results.map_estimates
```

```python
plot_corner(results)
plot_marginals(results, results.specs)
```
