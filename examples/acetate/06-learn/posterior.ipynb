# Posterior analysis

Inspect the result of `bff learn`. Run this notebook from `06-learn/`; it reads
`outputs/results.pt` and does not rerun any stage.

```python
from bff import Results

results = Results.load('outputs/results.pt')
results.summary()  # mean, std, quantiles, mode, and MAP of every parameter
```

## Plots

Marginals show the prior (grey), the posterior, the mode, and the MAP (diamond).

```python
fig = results.plot_marginals()
```

```python
fig = results.plot_corner()
```

## Validation parameters

`bff validate` draws parameter sets from `results.pt` itself (see
`07-validate/config.yaml`). To inspect or edit the sets first, export them and
pass the file as `parameters:` instead of `posterior:`.

```python
results.draw(
    10, include_mean=True, include_map=True, implicit=True,
    fn_out='outputs/posterior-samples.yaml', overwrite=True,
)
```
